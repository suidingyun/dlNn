# 使用Numpy实现拟合函数

```python
import numpy as np
import matplotlib.pyplot as plt


# 生成数据
x = np.linspace(-2 * np.pi, 2 * np.pi, 400)  # 输入范围 [-2π, 2π]
y = np.sin(x) - x  # 目标函数 sin(x) - x

# 使用 NumPy 多项式拟合
degree = 10  # 多项式的阶数
coefficients = np.polyfit(x, y, degree)  # 返回多项式的系数
polynomial = np.poly1d(coefficients)  # 构造多项式函数

# 使用拟合的多项式计算预测值
y_fit = polynomial(x)

# 绘图展示拟合效果
plt.plot(x, y, label="fact")  # 真实值曲线
plt.plot(x, y_fit, label="fit", linestyle="--")  # 拟合值曲线
plt.title("Fitting sin(x) - x with a Polynomial")
plt.xlabel("x")
plt.ylabel("sin(x) - x")
plt.legend()
plt.savefig(fname="Numpy_result.png", dpi=300)  # 保存图像
plt.show()
```
