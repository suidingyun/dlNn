# 基于神经网络拟合 sin(x)−x函数的实验报告

## 一、实验目的

本实验旨在通过两种不同的方法（NumPy 和 PyTorch）实现对函数 f(x)=sin(x)−x 的拟合，比较两种方法的特点和效果。具体目标包括：

- 掌握 NumPy 多项式拟合的基本原理和实现方法。
- 学习使用 PyTorch 构建神经网络模型进行非线性函数拟合。
- 对比两种方法在拟合精度、计算效率和适用场景上的差异。

## 二、函数定义

目标函数为：f(x)=sin(x)−x
该函数是一个非线性函数，具有以下特点：

- 包含周期性分量 
- sin(x)，呈现振荡特性。
- 线性分量 −x 导致整体趋势随 x 增大而减小。
综合了周期性和非线性特性，适合用于测试不同拟合方法的能力。

## 三、数据采集

### 1. 数据生成

为了训练和验证模型，我们在区间 [−2π,2π] 上均匀采样 400 个点，并计算对应的函数值。代码如下：

```python
import numpy as np

# 生成数据
x = np.linspace(-2 * np.pi, 2 * np.pi, 400)  # 输入范围 [-2π, 2π]
y = np.sin(x) - x  # 目标函数 sin(x) - x
```

### 2. 数据可视化

通过绘制散点图或曲线图，观察数据分布和函数特性：

```python
import matplotlib.pyplot as plt

plt.plot(x, y, label="fact")
plt.title("Data Distribution of sin(x) - x")
plt.xlabel("x")
plt.ylabel("sin(x) - x")
plt.legend()
plt.show()
```

## 四、模型描述

### 1. numpy多项式拟合

NumPy 提供了 numpy.polyfit 方法，可以基于最小二乘法拟合一个多项式函数。对于给定的数据集 
(x,y)，polyfit 返回一个多项式的系数向量。

实现步骤如下：
- 使用 np.polyfit(x, y, degree) 拟合一个指定阶数的多项式。
- 利用拟合得到的系数构造多项式函数
- 使用该多项式计算预测值，并与真实值对比。


代码实现：

```python
degree = 10  # 多项式的阶数
coefficients = np.polyfit(x, y, degree)  # 返回多项式的系数
polynomial = np.poly1d(coefficients)  # 构造多项式函数
y_fit_numpy = polynomial(x)  # 预测值
```

### 2. PyTorch 神经网络拟合

PyTorch 是一个强大的深度学习框架，可以通过构建神经网络模型来拟合复杂的非线性函数。神经网络通过多层非线性变换，能够逼近任意复杂的目标函数。


实现步骤
- 定义一个简单的全连接神经网络模型。
- 使用均方误差（MSE）作为损失函数，优化器选择 Adam。
- 训练模型，调整权重以最小化损失。


代码实现

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 转换数据格式
x_tensor = torch.tensor(x, dtype=torch.float32).unsqueeze(1)
y_tensor = torch.tensor(y, dtype=torch.float32).unsqueeze(1)

# 定义神经网络模型
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(1, 64)  # 输入层到隐藏层
        self.fc2 = nn.Linear(64, 64)  # 隐藏层到隐藏层
        self.fc3 = nn.Linear(64, 1)  # 隐藏层到输出层

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)
        return x

model = Net()

# 定义优化器和损失函数
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.MSELoss()

# 训练模型
epochs = 5000
for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(x_tensor)
    loss = criterion(outputs, y_tensor)
    loss.backward()
    optimizer.step()

# 预测
y_fit_torch = model(x_tensor).detach().numpy()
```
